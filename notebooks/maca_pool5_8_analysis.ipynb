```python
import pandas as pd
df2=pd.read_csv('test_5.csv')

split=df2["RuleVector"].str.split("-", expand=True)
df2["count_204"]=(split=="204").sum(axis=1)

df2.head()
```

```python
group_2=df2[df2["count_204"]==2]
group_2.head()
```

```python
size_2=group_2.shape[0]
size_2
```

```python
group_2_maca=group_2[group_2["MACA_or_not"]==1]
group_2_maca.head()
```

```python
size_2_ismaca=group_2_maca.shape[0]
size_2_ismaca
```

```python
mean=group_2["NumAttractors"].mean()
minimum=group_2["NumAttractors"].min()
maximum=group_2["NumAttractors"].max()
median=group_2["NumAttractors"].median()


print("Mean:", mean)
print("Minimum:", minimum) 
print("Maximum:", maximum)
print("Median:", median)
```
