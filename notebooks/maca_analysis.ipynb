```python
import pandas as pd
df1=pd.read_csv('test_4.csv')

split=df1["RuleVector"].str.split("-", expand=True)
df1["count_204"]=(split=="204").sum(axis=1)

df1.head()
```

```python
group_2=df1[df1["count_204"]==2]
group_2.head()
```

```python
size_2=group_2.shape[0]
size_2
```

```python
group_2_maca=group_2[group_2["MACA_or_not"]==1]
group_2_maca.head()
```

```python
size_2_ismaca=group_2_maca.shape[0]
size_2_ismaca
```

```python
mean=group_2["NumAttractors"].mean()
minimum=group_2["NumAttractors"].min()
maximum=group_2["NumAttractors"].max()
median=group_2["NumAttractors"].median()


print("Mean:", mean)
print("Minimum:", minimum) 
print("Maximum:", maximum)
print("Median:", median)
```
