```python
import pandas as pd
import matplotlib.pyplot as plt

df1=pd.read_csv('test_4.csv')
split=df1["RuleVector"].str.split("-", expand=True)
df1["count_204"]=(split=="204").sum(axis=1)

df1.head()
```

```python
count_values=[2,3,4,5,6]

mean=[]
median=[]

for val in count_values:
    subset=df1[df1["count_204"]==val]
    mean.append(subset["NumAttractors"].mean())
    median.append(subset["NumAttractors"].median())


x = range(len(count_values))
bar_width = 0.2

plt.figure(figsize=(12,6))

plt.bar([i - 0.5*bar_width for i in x], mean, width=bar_width, label="Mean", color="skyblue")

plt.bar([i + 0.5*bar_width for i in x], median, width=bar_width, label="Median", color="RED")

plt.xticks(x, count_values)
plt.legend()
plt.xlabel("Count of Rule 204 in RuleVector")
plt.ylabel("Attractors")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.title("Attractor analysis of Hybrid Rule Vectors with 4-self replicating RMTs as impurity")
plt.show()
```
