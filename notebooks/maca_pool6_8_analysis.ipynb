```python
import pandas as pd
df3=pd.read_csv('test_6.csv')

split=df3["RuleVector"].str.split("-", expand=True)
df3["count_204"]=(split=="204").sum(axis=1)

df3.head()
```

```python
group_2=df3[df3["count_204"]==2]
group_2.head()
```

```python
size_2=group_2.shape[0]
size_2
```

```python
mean=group_2["NumAttractors"].mean()
minimum=group_2["NumAttractors"].min()
maximum=group_2["NumAttractors"].max()
median=group_2["NumAttractors"].median()


print("Mean:", mean)
print("Minimum:", minimum) 
print("Maximum:", maximum)
print("Median:", median)
```
